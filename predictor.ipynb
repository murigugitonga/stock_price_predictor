```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pandas_datareader as data

plt.style.use('fivethirtyeight')
%matplotlib.inline
```

```python
import yfinance as yf
import datetime as dt

start = dt.datetime(2000, 1, 1)
end = dt.datetime.now()
stock = "OPAI.PVT" # OpenAI

df = yf.download(stock, start, end, True)
df
```

```python
df = df.reset_index()
```

```python
dataFOne= df.to_csv('opai.csv')
```

```python
dataFOne = pd.read_csv('opai.csv')
```

```python
dataFOne.head()
```

```python
# visualizing candlesticks
import plotly.graph_objects as go

fig = go.Figure(data=[go.Candlestick(x=dataFOne['Date'],
                open=dataFOne['Open'],
                high=dataFOne['High'],
                low=dataFOne['Low'],
                close=dataFOne['Close'])])

fig.update_layout(xaxis_rangeslider_visible=False)
fig.show()
```

```python
# removing unnecessary columns
df = df.drop(['Date'], axis=1)
```

```python
df.head()
```

```python
plt.figure(figsize=(12, 6))
#plt.grid(True)
plt.xlabel('Date')
plt.ylabel('Opening Price(USD)')
plt.plot(df['Open'], label = f"{stock} opening price", linewidth=1)
plt.title(f"{stock} closing price over time")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
#plt.grid(True)
plt.xlabel('Date')
plt.ylabel('Closing Price(USD)')
plt.plot(df['Close'], label = f"{stock} closing price", linewidth=1)
plt.title(f"{stock} closing price over time")
plt.legend()
plt.show()
## Same trend for high, low and volume
```
